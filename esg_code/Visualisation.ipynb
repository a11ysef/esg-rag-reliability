
## Introduction to Radar Chart for ESG Ratings Comparison

Environmental, Social, and Governance (ESG) ratings are essential metrics used to evaluate a company's sustainability and ethical impact. Different rating providers, such as MSCI, Sustainalytics, and S&P Global, offer their assessments, which can sometimes diverge significantly. To visualize these differences, a radar chart (also known as a spider chart) is an effective tool.

A radar chart allows us to plot multiple variables for different entities on a single graph, making it easy to compare their performance across various dimensions. In this notebook, we will use Plotly to create a radar chart that compares the ESG ratings of two companies—Tencent, Meta—across two rating providers: MSCI, Sustainalytics and S&P Global.

The radar chart will help us identify areas where the ratings diverge and provide a clear visual representation of each company's strengths and weaknesses according to different rating providers.

To visualize the ESG ratings and their divergence for different groups, we will create a series of plots. Each group, defined by the divergence range, will have its own set of subplots. Specifically, we will generate one radar plot for each ESG rating provider (MSCI and S&P Global) and a scatter plot for the divergence. All subplots will share the same legend for consistency and ease of comparison. The divergence groups are categorized as follows: low (1-3), middle (4-6), and high (7-10). This approach will help us understand the ESG ratings and their divergence across different companies and rating providers.


```python
from plotly.subplots import make_subplots

import plotly.graph_objects as go
import plotly.express as px
import os

def plot_radar_chart(df):
    return px.line_polar(
        df, 
        r="score", 
        theta="dimension", 
        color="Company", 
        line_close=True
        )  


def plot_rating_divergence(
        sector,
        industry,
        msci, 
        spglobal, 
        divergence, 
        divergence_min, 
        divergence_max
        ):
    companies = divergence[divergence.apply(
        lambda x: divergence_min<=x['divergence']<=divergence_max, axis=1
        )]['company']
    
    if not companies.any():
        print('No companies found in the given divergence range.')
        return
    df_msci = msci[msci['Company'].isin(companies)]
    df_spglobal = spglobal[spglobal['Company'].isin(companies)]
    df_div = divergence[divergence['company'].isin(companies)]
    # Create separate radar charts for MSCI indicators and S&P dimensions
    fig = make_subplots(rows=2, cols=2, 
                        specs=[[{'type': 'polar'}, {'type': 'polar'}],
                                [ {'type': 'scatter', "colspan": 2}, None]],
                        subplot_titles=("MSCI Indicators", "S&P Dimensions"))

    # Add MSCI indicators trace
    fig11 = plot_radar_chart(df_msci)
    for trace in fig11.data:
        fig.add_trace(trace, row=1, col=1)

    fig12 = plot_radar_chart(df_spglobal)
    for trace in fig12.data:
        fig.add_trace(trace, row=1, col=2)

    fig.update_traces(legendgroup='group1',showlegend=False, row=1, col=2)
                   
    fig13 = px.scatter(df_div, x='company', y='divergence', size='divergence', color='company')
    for trace in fig13.data:
        fig.add_trace(trace, row=2, col=1)
    fig.update_traces(legendgroup='group1',showlegend=False, row=2, col=1)

    # Customize the layout
    fig.update_layout(
        title='ESG Ratings Comparison (Spider Chart)',
        title_font_size=20,
        title_x=0.5,  # Center the title
        showlegend=True,
        legend=dict(
            orientation="h",  # Horizontal legend
            yanchor="bottom",
            y=1.02,
            xanchor="right",
            x=1
        ),
        height=800 , # Increase the height of the plot
        width=800  # Increase the width of the plot
    )
    folder = f'/Users/chenwanqiu/Downloads/llm_esg_judge-2_/submission/Task3_Visualisation/{sector}/{industry}'
    os.makedirs(folder, exist_ok=True)
    fig.write_html(f'{folder}/esg_ratings_comparison_divergence_{divergence_min}_{divergence_max}.html')
    # Show the chart
    fig.show()


sector = "Technology"
industry = "Internet Services and Retailing"
for div_min, div_max in [(1, 3), (4, 6), (7, 10)]:
    plot_rating_divergence(sector, industry, df_msci, df_spglobal, df_div, div_min, div_max)
```
